```python
import pandas as pd

pd.plotting.register_matplotlib_converters()

import matplotlib.pyplot as plt

%matplotlib inline

import seaborn as sns

print("Setup Complete")
```

```python
my_filepath = "/kaggle/input/datasets/debayank2024/netflix-movies-and-series/netflix_titles.csv"
```

```python

my_data = pd.read_csv(my_filepath)
```

```python
my_data.head()
```

```python
my_data.columns
```

```python
plt.figure(figsize=(8,5))

sns.countplot(data=my_data, x="type")

plt.title("Netflix Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=my_data,
    x="rating",
    order=my_data["rating"].value_counts().index
)

plt.title("Netflix Content by Rating")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.histplot(data=my_data, x="release_year", bins=30)

plt.title("Distribution of Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
plt.figure(figsize=(12,6))

sns.histplot(data=my_data, x="release_year", bins=30)

plt.title("Distribution of Netflix Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```
